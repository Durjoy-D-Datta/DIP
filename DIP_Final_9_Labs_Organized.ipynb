# Problem 1: Basic Gray Level Transformations
### Negative, Log Transformation and Power-Law (Gamma) Transformation


```python
img = cv2.imread("image.jpg", 0)

negative = np.zeros(img.shape, dtype=np.uint8)
logtrans = np.zeros(img.shape, dtype=np.uint8)
gamma_img = np.zeros(img.shape, dtype=np.uint8)

c = 255 / math.log(256)
g = 2

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        r = int(img[i][j])

        # Negative
        negative[i][j] = 255 - r

        # Log transformation
        s = c * math.log(1 + r)
        logtrans[i][j] = int(s)

        # Gamma transformation
        s = 255 * ((r / 255) ** g)
        gamma_img[i][j] = int(s)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(negative, cmap="gray")
plt.title("Negative")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(logtrans, cmap="gray")
plt.title("Log Transformation")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(gamma_img, cmap="gray")
plt.title("Gamma Transformation")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# Problem 2: Contrast Stretching and Gray Level Slicing


```python
img = cv2.imread("image.jpg", 0)

low = 50
high = 200

contrast = np.zeros(img.shape, dtype=np.uint8)
sliced = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        r = int(img[i][j])

        # Contrast stretching
        if r <= low:
            s = 0
        elif r >= high:
            s = 255
        else:
            s = ((r - low) * 255) / (high - low)

        contrast[i][j] = int(s)

        # Gray level slicing
        if low <= r <= high:
            sliced[i][j] = 255
        else:
            sliced[i][j] = 0

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(contrast, cmap="gray")
plt.title("Contrast Stretching")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sliced, cmap="gray")
plt.title("Gray Level Slicing")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# Problem 4: Histogram Equalization
### Manual histogram, CDF and equalization


```python
img = cv2.imread("image.jpg", 0)

# Original histogram
histogram = np.zeros(256, dtype=int)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        r = int(img[i][j])
        histogram[r] += 1

# CDF
cdf = np.zeros(256, dtype=int)
cdf[0] = histogram[0]

for i in range(1, 256):
    cdf[i] = cdf[i - 1] + histogram[i]

# First non-zero CDF
cdf_min = 0

for i in range(256):
    if cdf[i] != 0:
        cdf_min = cdf[i]
        break

total_pix = img.shape[0] * img.shape[1]

# Equalized image
equalized = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        r = int(img[i][j])

        s = ((cdf[r] - cdf_min) * 255) / (total_pix - cdf_min)

        if s < 0:
            s = 0
        if s > 255:
            s = 255

        equalized[i][j] = int(s)

# Equalized histogram
equalized_hist = np.zeros(256, dtype=int)

for i in range(equalized.shape[0]):
    for j in range(equalized.shape[1]):
        r = int(equalized[i][j])
        equalized_hist[r] += 1

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(equalized, cmap="gray")
plt.title("Equalized Image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.bar(range(256), histogram, width=1)
plt.title("Original Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 4)
plt.bar(range(256), equalized_hist, width=1)
plt.title("Equalized Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

# Problem 5: Arithmetic and Logical Operations
### Addition, Subtraction, Multiplication, Division, AND, OR, NOT and XOR


```python
img1 = cv2.imread("image1.jpg", 0)
img2 = cv2.imread("image2.jpg", 0)

add = np.zeros(img1.shape, dtype=np.uint8)
sub = np.zeros(img1.shape, dtype=np.uint8)
mul = np.zeros(img1.shape, dtype=np.uint8)
div = np.zeros(img1.shape, dtype=np.uint8)

and_img = np.zeros(img1.shape, dtype=np.uint8)
or_img = np.zeros(img1.shape, dtype=np.uint8)
not_img = np.zeros(img1.shape, dtype=np.uint8)
xor_img = np.zeros(img1.shape, dtype=np.uint8)

for i in range(img1.shape[0]):
    for j in range(img1.shape[1]):
        a = int(img1[i][j])
        b = int(img2[i][j])

        # Addition
        val = a + b
        val = min(val, 255)
        add[i][j] = int(val)

        # Subtraction
        val = a - b
        val = max(val, 0)
        sub[i][j] = int(val)

        # Multiplication
        val = (a * b) // 255
        mul[i][j] = int(val)

        # Division
        if b != 0:
            val = (a * 255) / b
            val = min(val, 255)
            div[i][j] = int(val)
        else:
            div[i][j] = 0

        # Logical / bitwise operations
        and_img[i][j] = a & b
        or_img[i][j] = a | b
        not_img[i][j] = 255 - a
        xor_img[i][j] = a ^ b

images = [add, sub, mul, div, and_img, or_img, not_img, xor_img]
titles = ["Addition", "Subtraction", "Multiplication", "Division",
          "AND", "OR", "NOT", "XOR"]

plt.figure(figsize=(12, 12))

for i in range(8):
    plt.subplot(3, 3, i + 1)
    plt.imshow(images[i], cmap="gray")
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

# Problem 6: Bit Plane Slicing


```python
img = cv2.imread("image.jpg", 0)

planes = []

for k in range(8):
    plane = np.zeros(img.shape, dtype=np.uint8)

    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            r = int(img[i][j])
            bit = (r >> k) & 1
            plane[i][j] = bit * 255

    planes.append(plane)

plt.figure(figsize=(12, 8))

for k in range(8):
    plt.subplot(2, 4, k + 1)
    plt.imshow(planes[k], cmap="gray")
    plt.title("Bit Plane " + str(k))
    plt.axis("off")

plt.tight_layout()
plt.show()
```

# Problem 7: Spatial Filtering
### Smoothing, Sharpening and Sobel Edge Detection


```python
img = cv2.imread("image.jpg", 0)

padded = np.pad(img, 1, mode="constant")

smooth = np.zeros(img.shape, dtype=np.uint8)
sharp = np.zeros(img.shape, dtype=np.uint8)
edge = np.zeros(img.shape, dtype=np.uint8)

mean_mask = np.ones((3, 3), dtype=np.float32) / 9

lap_mask = np.array([
    [0, 1, 0],
    [1, -4, 1],
    [0, 1, 0]
])

sobel_mx = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
])

sobel_my = np.array([
    [-1, -2, -1],
    [0, 0, 0],
    [1, 2, 1]
])

for i in range(img.shape[0]):
    for j in range(img.shape[1]):

        region = padded[i:i+3, j:j+3]

        # Smoothing
        smooth_val = np.sum(region * mean_mask)

        if smooth_val > 255:
            smooth_val = 255
        if smooth_val < 0:
            smooth_val = 0

        smooth[i][j] = int(smooth_val)

        # Laplacian sharpening
        lap_val = np.sum(region * lap_mask)
        sharp_val = int(img[i][j]) - int(lap_val)

        if sharp_val > 255:
            sharp_val = 255
        if sharp_val < 0:
            sharp_val = 0

        sharp[i][j] = int(sharp_val)

        # Sobel edge detection
        gx = np.sum(region * sobel_mx)
        gy = np.sum(region * sobel_my)

        edge_val = math.sqrt(gx * gx + gy * gy)

        if edge_val > 255:
            edge_val = 255
        if edge_val < 0:
            edge_val = 0

        edge[i][j] = int(edge_val)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(smooth, cmap="gray")
plt.title("Smoothed Image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(sharp, cmap="gray")
plt.title("Sharpened Image")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(edge, cmap="gray")
plt.title("Sobel Edge Detection")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# Problem 9: Basic Morphological Operations
### Dilation, Erosion, Opening and Closing


```python
img = cv2.imread("image.jpg", 0)

if img is None:
    raise FileNotFoundError("Image not found")

_, img = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

mask = np.ones((3, 3), dtype=np.uint8)

def morph(image, operation):

    padded = np.pad(image, 1, mode="constant")
    result = np.zeros(image.shape, dtype=np.uint8)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):

            region = padded[i:i+3, j:j+3]

            if operation == "dilation":
                result[i][j] = np.max(region * mask)

            elif operation == "erosion":
                result[i][j] = np.min(region * mask)

    return result

dilation = morph(img, "dilation")
erosion = morph(img, "erosion")

# Opening = Erosion -> Dilation
opening = morph(erosion, "dilation")

# Closing = Dilation -> Erosion
closing = morph(dilation, "erosion")

images = [img, dilation, erosion, opening, closing]
titles = ["Original Binary Image", "Dilation", "Erosion", "Opening", "Closing"]

plt.figure(figsize=(10, 6))

for i in range(5):
    plt.subplot(2, 3, i + 1)
    plt.imshow(images[i], cmap="gray")
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```
